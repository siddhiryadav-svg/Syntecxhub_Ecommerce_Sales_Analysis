# E-commerce Sales Data Analysis

## Internship Project – Syntecxhub

### Project Objective
To read, inspect, analyse and filter e-commerce sales data using Python Pandas.

### Dataset Information
- Total Records: 1000
- Data Type: Synthetic E-commerce Sales Data
- Tools: Python, Pandas, NumPy, Matplotlib
- File Format: CSV and Excel

```python
import pandas as pd
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

# Number of records
n = 1000

# Create dataset
data = {
    "Order_ID": [f"ORD{i:04d}" for i in range(1, n + 1)],
    "Order_Date": pd.date_range(
        start="2025-01-01", periods=n, freq="D"
    ),
    "Customer_ID": np.random.randint(1001, 1201, n),
    "Product": np.random.choice(
        ["Laptop", "Mobile", "Headphones", "Keyboard",
         "Mouse", "Monitor", "Tablet", "Smartwatch"], n
    ),
    "Category": np.random.choice(
        ["Electronics", "Accessories", "Computers"], n
    ),
    "Quantity": np.random.randint(1, 6, n),
    "Unit_Price": np.random.choice(
        [500, 1000, 1500, 2000, 5000, 10000, 20000, 40000], n
    ),
    "Payment_Method": np.random.choice(
        ["UPI", "Credit Card", "Debit Card", "Cash on Delivery"], n
    ),
    "City": np.random.choice(
        ["Mumbai", "Delhi", "Pune", "Bangalore",
         "Chennai", "Hyderabad", "Jaipur", "Kolkata"], n
    )
}

# Create DataFrame
df = pd.DataFrame(data)

# Calculate total sales
df["Total_Sales"] = df["Quantity"] * df["Unit_Price"]

# Save dataset
df.to_csv("ecommerce_dataset.csv", index=False)

print("Dataset created successfully!")
print("Total Records:", len(df))
```

```python
product_categories = {
    "Laptop": "Computers",
    "Mobile": "Electronics",
    "Headphones": "Accessories",
    "Keyboard": "Accessories",
    "Mouse": "Accessories",
    "Monitor": "Computers",
    "Tablet": "Electronics",
    "Smartwatch": "Electronics"
}
```

```python
df["Category"] = df["Product"].map(product_categories)

print("Product categories corrected!")
display(df[["Product", "Category"]].head(15))
```

```python
print(df.groupby(["Product", "Category"]).size())
```

```python
product_prices = {
    "Laptop": (35000, 80000),
    "Mobile": (10000, 50000),
    "Headphones": (500, 5000),
    "Keyboard": (500, 3000),
    "Mouse": (300, 2000),
    "Monitor": (5000, 25000),
    "Tablet": (10000, 40000),
    "Smartwatch": (1500, 12000)
}

df["Unit_Price"] = df["Product"].apply(
    lambda product: np.random.randint(
        product_prices[product][0],
        product_prices[product][1] + 1
    )
)

df["Total_Sales"] = df["Quantity"] * df["Unit_Price"]

print("Product prices updated successfully!")

display(df[["Product", "Unit_Price", "Quantity", "Total_Sales"]].head(10))
```

```python
print("Total Duplicate Rows:", df.duplicated().sum())
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Invalid Quantity:", (df["Quantity"] <= 0).sum())

print("Invalid Unit Price:", (df["Unit_Price"] <= 0).sum())

print("Incorrect Total Sales:",
      (df["Total_Sales"] != df["Quantity"] * df["Unit_Price"]).sum())
```

```python
# Save the improved dataset
df.to_csv("ecommerce_dataset.csv", index=False)

# Save complete dataset in Excel format
df.to_excel("ecommerce_dataset.xlsx", index=False)

print("Final dataset saved successfully!")
print("Total Records:", len(df))
```

```python
import matplotlib.pyplot as plt

category_sales = df.groupby("Category")["Total_Sales"].sum()

category_sales.plot(kind="bar", figsize=(8, 5))

plt.title("Category-wise Sales")
plt.xlabel("Category")
plt.ylabel("Total Sales (₹)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
city_sales = df.groupby("City")["Total_Sales"].sum()

city_sales.plot(kind="bar", figsize=(9, 5))

plt.title("City-wise Sales")
plt.xlabel("City")
plt.ylabel("Total Sales (₹)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
top_products = (
    df.groupby("Product")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

top_products.plot(kind="bar", figsize=(8, 5))

plt.title("Top 5 Products by Revenue")
plt.xlabel("Product")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```
